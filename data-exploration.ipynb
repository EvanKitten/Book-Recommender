```python
import kagglehub
from matplotlib.pyplot import plot_date

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
books = pd.read_csv(f"{path}/books.csv")
```

```python
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
#Heatmap documents all of the missing values
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing Values")
plt.show()
```

```python
import numpy as np

#books where description is missing put 1, else put 0
books["missing_description"] = np.where(books["description"].isna(), 1, 0)
#convert year of book to age of book
books["age_of_book"] = 2025 - books["published_year"]
```

```python
#create a list with just the subset of columns that we want from our books dataset
columns_of_interest = ["num_pages", "age_of_book", "missing_description", "average_rating"]

#create correlation matrix off of pandas dataframe
#spearman over pearson, pearson - completely continuous variables, correlation is binary so spearman for noncontinuous values
correlation_matrix = books[columns_of_interest].corr(method = "spearman")

sns.set_theme(style="white")
plt.figure(figsize = (8,6))
heatmap = sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar_kws={"label": " Spearman Correlation"})

heatmap.set_title("Correlation Heatmap")
plt.show()

#Want to see correlation with missing description, due to bias being based off of missing description
#Anything close to 1 or -1 is strong correlation, closer to 0 weak correlation
```

```python
#If variable is missing, fetch me those records
books[(books["description"].isna()) |
      (books["num_pages"].isna()) |
      (books["average_rating"].isna()) |
      (books["published_year"].isna())]
```

```python
# ~ means negation
#only books with variable as complete values
book_missing = books[~(books["description"].isna()) &
      ~(books["num_pages"].isna()) &
      ~(books["average_rating"].isna()) &
      ~(books["published_year"].isna())]

book_missing
```

```python
#Look at distribution of categories by taking value count of each
#reset index so that categories are put back in a column, most popular at top
book_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
book_missing
```

```python
#take description, split based on space (default for str.split()), take the length, gives us count of words
#book_missing.loc[:,"words_in_description"] = book_missing["description"].str.split().str.len()
book_missing = book_missing.copy()

# Now safely assign new column
book_missing["words_in_description"] = book_missing["description"].str.split().str.len()
```

```python
book_missing
```

```python
#filter so that description is between 1 and 4 words
book_missing.loc[book_missing["words_in_description"].between(1,4), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(5,14), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(15,24), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(25,34), "description"]
```

```python
#Keep descriptions with more than 25 words
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25]
book_missing_25_words
```

```python
#if subtitle is missing, just use title, else aggregate subtitle and title. Joined using colon
book_missing_25_words = book_missing_25_words.copy()

book_missing_25_words["title_and_subtitle"] = np.where(
    book_missing_25_words["subtitle"].isna(),
    book_missing_25_words["title"],
    book_missing_25_words[["title", "subtitle"]].astype(str).agg(": ".join, axis=1)
)
```

```python
book_missing_25_words
```

```python
#isbn attached as first word of description
book_missing_25_words.loc[:,"tagged_description"] = book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
book_missing_25_words
```

```python
#dropping columns that are not needed
(
    book_missing_25_words
    .drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
    .to_csv("books_cleaned.csv", index=False)
)
```
